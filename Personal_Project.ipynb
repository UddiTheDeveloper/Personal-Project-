```python
# Extracting form Zip file
import zipfile
zip_extract= zipfile.ZipFile('archive.zip')
zip_extract.extractall('extract_folder')
zip_extract.close()
```

```python
# Loading and displaying Dataset
import pandas as pd
Data= pd.read_csv('/content/extract_folder/track_data_final.csv')
Data.head()

```

```python
print("Basic Information:-")
Data.info()
```

```python
print("Null Values:-")
Data.isnull().sum()
```

```python
Data.dropna(inplace=True)
Data.isnull().sum()
```

```python
import pandas as pd
Data['artist_popularity'] = Data['artist_popularity'].fillna(0)
Data['artist_popularity'] = Data['artist_popularity'].astype(int)
Data['artist_popularity'].unique()
Data['album_release_date'] = pd.to_datetime(Data['album_release_date'], errors='coerce')
Data.dtypes
```

```python
#finding ou duplicates
Data.duplicated().sum()
```

```python
# Removing Duplicates
Data.drop_duplicates(inplace=True)
```

```python

Data['explicit'].value_counts()
```

```python
Data['album_type'].value_counts()
```

```python
Data['artist_genres'].value_counts()
```

```python
print("Statistical Analysis:-")
round(Data.describe())
```

```python
#Removing Outliers:-
Data= Data[Data['track_number']!=102.0]
Data=Data[Data['album_total_tracks']!=181.0]
```

```python
print("Average no of tracks per type group by analysis:-")
album_per_artist= Data.groupby('album_type')['album_total_tracks'].mean()
round(album_per_artist)
```

```python
Data.to_csv('track_data_final.csv', index=False)
```
