```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Define the numerical features to plot
features = ['Price_USD', 'Units_Sold', 'Revenue_USD']
feature_colors = ['#2b5c8f', '#d95f02', '#2ca02c']

# Create subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

for i, (col, color) in enumerate(zip(features, feature_colors)):
    # Draw boxplot on the specific axis
    sns.boxplot(y=df[col], ax=axes[i], color=color, width=0.4, flierprops={'marker': 'o', 'markerfacecolor': 'red', 'markersize': 6})

    # Calculate summary statistics to use as data labels
    stats = df[col].describe()
    min_val = df[col].min()
    q1 = stats['25%']
    median = stats['50%']
    q3 = stats['75%']
    max_val = df[col].max()

    # Define annotation positions on the plots
    # We place annotations slightly offset to the right of the box (x = 0.25)
    x_pos = 0.25

    axes[i].text(x_pos, min_val, f"Min: {min_val:.2f}", color='black', va='center', ha='left', fontsize=9, fontweight='semibold')
    axes[i].text(x_pos, q1, f"Q1: {q1:.2f}", color='black', va='center', ha='left', fontsize=9, fontweight='semibold')
    axes[i].text(x_pos, median, f"Median: {median:.2f}", color='darkred', va='center', ha='left', fontsize=9.5, fontweight='bold')
    axes[i].text(x_pos, q3, f"Q3: {q3:.2f}", color='black', va='center', ha='left', fontsize=9, fontweight='semibold')
    axes[i].text(x_pos, max_val, f"Max: {max_val:.2f}", color='black', va='center', ha='left', fontsize=9, fontweight='semibold')

    # Styling improvements
    axes[i].set_title(f'Distribution of {col}', fontsize=12, fontweight='bold', pad=12)
    axes[i].set_ylabel(col, fontsize=10)
    axes[i].set_xlabel('')
    axes[i].grid(True, linestyle='--', alpha=0.4, axis='y')
    axes[i].set_xlim(-0.5, 0.8) # Extend limit to leave space for labels on the right

plt.suptitle('Boxplots with Descriptive Data Labels', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

Exploratory Data Analysis:-


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group the data by Country and Brand, summing the Revenue
revenue_df = df.groupby(['Country', 'Brand'])['Revenue_USD'].sum().unstack().fillna(0)

# Plot a smaller stacked bar chart with darker colors
fig, ax = plt.subplots(figsize=(10, 6))
revenue_df.plot(kind='bar', stacked=True, ax=ax, colormap='magma')

# Add data labels inside the stacks with customized text colors for readability
for i, c in enumerate(ax.containers):
    # Identify the brand name for this container
    brand_name = revenue_df.columns[i]

    # Use a dark text color for Skechers (which is a very light shade in 'magma') for better contrast
    text_color = '#1a1a1a' if brand_name == 'Skechers' else 'white'

    # Only label segments that have a visible height to avoid cluttering with zero labels
    labels = [f"${val/1000:.1f}k" if val > 5000 else "" for val in c.datavalues]
    ax.bar_label(c, labels=labels, label_type='center', fontsize=8, color=text_color, fontweight='bold')

# Set labels and title
plt.title('Total Revenue per Brand by Country', fontsize=12, fontweight='bold')
plt.xlabel('Country', fontsize=10)
plt.ylabel('Total Revenue (USD)', fontsize=10)
plt.xticks(rotation=45, fontsize=9)
plt.yticks(fontsize=9)
plt.legend(title='Brand', bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=9)
plt.tight_layout()

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import math

# Group data by Country and Sales Channel, then calculate sum of Units Sold
channel_units = df.groupby(['Country', 'Sales_Channel'])['Units_Sold'].sum().unstack().fillna(0)

# Define the grid layout for the subplots
countries = channel_units.index
num_countries = len(countries)
cols = 3
rows = math.ceil(num_countries / cols)

# Set up the figure with clean dimensions
fig, axes = plt.subplots(rows, cols, figsize=(15, 5 * rows))
axes = axes.flatten()  # Flatten the 2D array of axes for easy indexing

# Define colors for consistency across all pie charts
colors = sns.color_palette('pastel')[0:len(channel_units.columns)]

for i, country in enumerate(countries):
    country_data = channel_units.loc[country]

    # Plot pie chart on the specific subplot axis
    axes[i].pie(
        country_data,
        labels=country_data.index,
        autopct='%1.1f%%',
        startangle=140,
        colors=colors,
        textprops={'fontsize': 10, 'weight': 'bold'}
    )
    axes[i].set_title(f'Sales Channel Mix: {country}', fontsize=12, fontweight='bold', pad=15)

# Hide any unused axes in the grid
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.suptitle('Percentage of Units Sold per Sales Channel by Country', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the Pearson correlation coefficient
correlation = df['Price_USD'].corr(df['Units_Sold'])
print(f"Pearson correlation coefficient: {correlation:.4f}")

# Plot a smaller scatter plot with a regression line
plt.figure(figsize=(7, 4.5))
sns.regplot(
    data=df,
    x='Price_USD',
    y='Units_Sold',
    scatter_kws={'alpha':0.4, 'color': '#2b5c8f'},
    line_kws={'color': '#d95f02', 'linewidth': 2}
)

plt.title(f'Correlation between Price and Units Sold (r = {correlation:.3f})', fontsize=11, fontweight='bold')
plt.xlabel('Price (USD)', fontsize=10)
plt.ylabel('Units Sold', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style
sns.set_theme(style="whitegrid")

# Create subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# 1. Average Sales by Shoe Type
shoe_type_avg = df.groupby('Shoe_Type')['Units_Sold'].mean().sort_values(ascending=False).reset_index()
sns.barplot(x='Shoe_Type', y='Units_Sold', data=shoe_type_avg, ax=axes[0], hue='Shoe_Type', palette='Blues_r', legend=False)
axes[0].set_title('Average Units Sold by Shoe Type', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Shoe Type')
axes[0].set_ylabel('Average Units Sold')
axes[0].tick_params(axis='x', rotation=45)
# Add data labels
for container in axes[0].containers:
    axes[0].bar_label(container, fmt='%.2f', padding=3, fontsize=9, fontweight='semibold')

# 2. Average Sales by Brand
brand_avg = df.groupby('Brand')['Units_Sold'].mean().sort_values(ascending=False).reset_index()
sns.barplot(x='Brand', y='Units_Sold', data=brand_avg, ax=axes[1], hue='Brand', palette='Oranges_r', legend=False)
axes[1].set_title('Average Units Sold by Brand', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Brand')
axes[1].set_ylabel('')
axes[1].tick_params(axis='x', rotation=45)
# Add data labels
for container in axes[1].containers:
    axes[1].bar_label(container, fmt='%.2f', padding=3, fontsize=9, fontweight='semibold')

# 3. Average Sales by Sales Channel
channel_avg = df.groupby('Sales_Channel')['Units_Sold'].mean().sort_values(ascending=False).reset_index()
sns.barplot(x='Sales_Channel', y='Units_Sold', data=channel_avg, ax=axes[2], hue='Sales_Channel', palette='Greens_r', legend=False)
axes[2].set_title('Average Units Sold by Sales Channel', fontsize=12, fontweight='bold')
axes[2].set_xlabel('Sales Channel')
axes[2].set_ylabel('')
axes[2].tick_params(axis='x', rotation=45)
# Add data labels
for container in axes[2].containers:
    axes[2].bar_label(container, fmt='%.2f', padding=3, fontsize=9, fontweight='semibold')

# Overall Layout Tweaks
plt.suptitle('Comparison of Average Sales (Units Sold)', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
